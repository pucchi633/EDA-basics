# Q1. What is the average, median, and standard deviation of house prices?


```python
avg_price = df['Price'].mean()
median_price = df['Price'].median()
std_price = df['Price'].std()

print("Average House Price =", avg_price)
print("Median House Price =", median_price)
print("Standard Deviation =", std_price)


```

#Q2. Which number of bedrooms is most common?


```python
most_common_bedrooms = df['number of bedrooms'].mode()[0]

print("Most Common Number of Bedrooms =", most_common_bedrooms)
```

#Q3. Check if the price data is skewed

```python
skewness = df['Price'].skew()

print("Skewness of Price =", skewness)
```

Skewness is used to measure the distribution of house prices.

If the skewness value is positive, the data is right-skewed, meaning a few high-priced houses increase the average price.

If the skewness value is close to zero, the data is approximately normally distributed.

Therefore, the skewness value helps understand the overall pattern of house prices in the dataset.

#Q4. What is the average price for each number of bedrooms?


```python
avg_price_bedrooms = df.groupby('number of bedrooms')['Price'].mean()

print(avg_price_bedrooms)
```

#Q5. What is the relationship between living area and price? (Use : Area without Basement).


```python
df[['Area of the house(excluding basement)', 'Price']].corr()
```

# Q6. Identify any anomalies where houses have high prices but low areas.

```python
anomalies = df[
    (df['Price'] > df['Price'].quantile(0.90)) &
    (df['Area of the house(excluding basement)'] < df['Area of the house(excluding basement)'].quantile(0.10))
]

anomalies[['Price', 'Area of the house(excluding basement)']]
```

# Q7. Compare average price based on number of floors and houses with or without waterfront.

```python
avg_price_comparison = df.groupby(['number of floors', 'waterfront present'])['Price'].mean()

print(avg_price_comparison)
```

# Q8. Identify the minimum and maximum house price. What does this indicate?

```python
min_price = df['Price'].min()
max_price = df['Price'].max()

print("Minimum House Price =", min_price)
print("Maximum House Price =", max_price)
```

# Q9. Which location (zipcode/area) has the highest average price?


```python
highest_avg_price_location = df.groupby('Postal Code')['Price'].mean().sort_values(ascending=False).head(1)

print(highest_avg_price_location)
```
